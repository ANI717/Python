```python
import sys
import traceback
```

### Dictionary

```python
# Create and print a dictionary in Python

mydict = {"name": "John", "age": 30, "city": "New York"}
print(mydict)
```

```python
# Create and print a dictionary using the dict() constructor in Python

mydict = dict(name="John", age=30, city="New York")
print(mydict)
```

### Access Values

```python
# Access and print dictionary values by key in Python

print(mydict["name"])
print(mydict["age"])
print(mydict["city"])   
```

```python
# Handle any exception when accessing a dictionary key in Python

try:
    print(mydict["first_name"])
except Exception as ex:
    traceback.print_exc()
```

### Insert

```python
# Add a new key-value pair to the dictionary in Python

mydict["email"] = "john@example.com"
print(mydict)
```

### Remove

```python
# Delete a key-value pair from the dictionary in Python

del mydict["email"]
print(mydict)
```

```python
# Remove and return a key-value pair from the dictionary in Python

value = mydict.pop("city")
print(value)
print(mydict)
```

### Check item

```python
# Check if a key exists in the dictionary in Python

print("name" in mydict)
print("email" in mydict)
print("city" in mydict)
```

### Iterate

```python
# Iterate over all key-value pairs in the dictionary and print them

for key, value in mydict.items():
    print(f"{key}: {value}")
```

```python
# Iterate over all keys in the dictionary and print them

for key in mydict.keys():
    print(key)
```

```python
# Iterate over all values in the dictionary and print them

for value in mydict.values():
    print(value)
```

### Copy

```python
# Demonstrate that modifying a copy of a dictionary affects the original dictionary when using assignment

mydict_copy = mydict
mydict_copy["name"] = "Max"
print(mydict)
print(mydict_copy)
```

```python
# Demonstrate that modifying a copy of a dictionary does not affect the original dictionary when using the copy() method

mydict_copy = mydict.copy()
mydict_copy["name"] = "Ani"
print(mydict)
print(mydict_copy)
```

```python
# Demonstrate that modifying a copy of a dictionary created using the dict() constructor does not affect the original dictionary

mydict_copy = dict(mydict)
mydict_copy["name"] = "John"
print(mydict)
print(mydict_copy)
```

### Merge

```python
dict1 = {"name": "Alice", "age": 25, "city": "New York"}
dict2 = {"height": 170, "weight": 60}
```

```python
# Demonstrate updating a dictionary with another dictionary using the update() method

dict1.update(dict2)
print(dict1)
```
