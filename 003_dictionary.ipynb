```python
import sys
import traceback
```

### Dictionary

```python
# Create and print a dictionary in Python

mydict = {"name": "John", "age": 30, "city": "New York"}
print(mydict)
```

```python
# Create and print a dictionary using the dict() constructor in Python

mydict = dict(name="John", age=30, city="New York")
print(mydict)
```

```python
# Create an empty dictionary; {} is a dictionary, so an empty set needs set()

empty1 = {}
empty2 = dict()
print(empty1, empty2)
print(type(empty1))
print(type(set()))
```

```python
# Create a dictionary from a list of pairs and from two lists using zip

pairs = dict([("name", "John"), ("age", 30)])
print(pairs)

keys = ["name", "age"]
values = ["John", 30]
print(dict(zip(keys, values)))
```

```python
# Create a dictionary with the same value for every key using fromkeys

print(dict.fromkeys(["a", "b", "c"], 0))
print(dict.fromkeys(["a", "b", "c"]))
```

### Keys

```python
# If a key is repeated, the last value wins

repeated = {"name": "John", "age": 30, "name": "Max"}
print(repeated)
```

```python
# 1, 1.0 and True are equal and have the same hash, so they count as the same key

same_key = {1: "int", 1.0: "float", True: "bool"}
print(same_key)
```

```python
# A tuple can be used as a key because it is hashable

locations = {(39.95, -75.16): "Philly", (40.71, -74.01): "New York"}
print(locations[(39.95, -75.16)])
```

```python
# A list cannot be used as a key because it is not hashable

try:
    locations = {[39.95, -75.16]: "Philly"}
except Exception as ex:
    traceback.print_exc()
```

### Access Values

```python
# Access and print dictionary values by key in Python

print(mydict["name"])
print(mydict["age"])
print(mydict["city"])   
```

```python
# Handle any exception when accessing a dictionary key in Python

try:
    print(mydict["first_name"])
except Exception as ex:
    traceback.print_exc()
```

```python
# Use get() to access a key without raising an error when it is missing

person = {"name": "John", "age": 30}
print(person.get("name"))
print(person.get("email"))
print(person.get("email", "not provided"))
```

### Insert

```python
# Add a new key-value pair to the dictionary in Python

mydict["email"] = "john@example.com"
print(mydict)
```

```python
# Assigning to an existing key overwrites its value

person = {"name": "John", "age": 30}
person["age"] = 31
print(person)
```

```python
# setdefault() adds the key only if it is missing and returns the current value

person = {"name": "John", "age": 30}
print(person.setdefault("age", 99))
print(person.setdefault("city", "Philly"))
print(person)
```

### Remove

```python
# Delete a key-value pair from the dictionary in Python

del mydict["email"]
print(mydict)
```

```python
# Remove and return a key-value pair from the dictionary in Python

value = mydict.pop("city")
print(value)
print(mydict)
```

```python
# pop() with a default does not raise an error when the key is missing

person = {"name": "John", "age": 30}
print(person.pop("email", "not found"))
```

```python
# pop() without a default raises an error when the key is missing

try:
    person.pop("email")
except Exception as ex:
    traceback.print_exc()
```

```python
# del also raises an error when the key is missing

try:
    del person["email"]
except Exception as ex:
    traceback.print_exc()
```

```python
# popitem() removes and returns the last inserted key-value pair

person = {"name": "John", "age": 30, "city": "Philly"}
print(person.popitem())
print(person)
```

```python
# clear() removes all key-value pairs

person.clear()
print(person)
```

### Check item

```python
# Check if a key exists in the dictionary in Python

print("name" in mydict)
print("email" in mydict)
print("city" in mydict)
```

```python
# in only checks keys, so use values() to check for a value

person = {"name": "John", "age": 30}
print("John" in person)
print("John" in person.values())
```

```python
# Get the number of key-value pairs in the dictionary

print(len(person))
```

### Iterate

```python
# Iterate over all key-value pairs in the dictionary and print them

for key, value in mydict.items():
    print(f"{key}: {value}")
```

```python
# Iterate over all keys in the dictionary and print them

for key in mydict.keys():
    print(key)
```

```python
# Iterate over all values in the dictionary and print them

for value in mydict.values():
    print(value)
```

```python
# A dictionary keeps its insertion order

letters = {}
letters["z"] = 1
letters["a"] = 2
letters["m"] = 3
print(list(letters))
```

```python
# keys(), values() and items() are live views that reflect later changes

person = {"name": "John", "age": 30}
person_keys = person.keys()
print(person_keys)

person["city"] = "Philly"
print(person_keys)
```

```python
# Adding a key while iterating over a dictionary raises an error

scores = {"a": 1, "b": 2}
try:
    for key in scores:
        scores[key + "_new"] = 0
except Exception as ex:
    traceback.print_exc()
```

```python
# Iterating over a list of the keys avoids the problem

scores = {"a": 1, "b": 2}
for key in list(scores):
    scores[key + "_new"] = 0
print(scores)
```

### Ordering

```python
# sorted() on a dictionary returns a sorted list of its keys

scores = {"Charlie": 70, "Alice": 90, "Bob": 80}
print(sorted(scores))
```

```python
# Sort a dictionary by value using a key function

print(sorted(scores.items(), key=lambda item: item[1]))
print(dict(sorted(scores.items(), key=lambda item: item[1])))
```

```python
# Comparing dictionaries with == ignores the order

print({"a": 1, "b": 2} == {"b": 2, "a": 1})
```

### Copy

```python
# Demonstrate that modifying a copy of a dictionary affects the original dictionary when using assignment

mydict_copy = mydict
mydict_copy["name"] = "Max"
print(mydict)
print(mydict_copy)
```

```python
# Demonstrate that modifying a copy of a dictionary does not affect the original dictionary when using the copy() method

mydict_copy = mydict.copy()
mydict_copy["name"] = "Ani"
print(mydict)
print(mydict_copy)
```

```python
# Demonstrate that modifying a copy of a dictionary created using the dict() constructor does not affect the original dictionary

mydict_copy = dict(mydict)
mydict_copy["name"] = "John"
print(mydict)
print(mydict_copy)
```

```python
# copy() makes a shallow copy, so a nested list is still shared

student = {"name": "John", "grades": [90, 80]}
shallow = student.copy()
shallow["grades"].append(70)
print(student)
print(shallow)
```

```python
# deepcopy() copies the nested list too

import copy

student = {"name": "John", "grades": [90, 80]}
deep = copy.deepcopy(student)
deep["grades"].append(70)
print(student)
print(deep)
```

### Merge

```python
dict1 = {"name": "Alice", "age": 25, "city": "New York"}
dict2 = {"height": 170, "weight": 60}
```

```python
# Demonstrate updating a dictionary with another dictionary using the update() method

dict1.update(dict2)
print(dict1)
```

```python
# When both dictionaries have the same key, the value from the right side wins

left = {"name": "Alice", "age": 25}
right = {"age": 30, "city": "Philly"}

merged = left.copy()
merged.update(right)
print(merged)
```

```python
# The | operator creates a new merged dictionary and leaves the originals unchanged

print(left | right)
print(left)
```

```python
# The |= operator merges into the existing dictionary

merged = left.copy()
merged |= right
print(merged)
```

```python
# Unpacking with ** also creates a new merged dictionary

print({**left, **right})
```

### Dictionary Comprehension

```python
# Create a dictionary using a comprehension

squares = {x: x * x for x in range(5)}
print(squares)
```

```python
# Create a dictionary using a comprehension with a condition (filter)

scores = {"Alice": 90, "Bob": 60, "Charlie": 75}
passed = {name: score for name, score in scores.items() if score >= 70}
print(passed)
```

```python
# Swap keys and values using a comprehension

print({score: name for name, score in scores.items()})
```

### Nested dictionaries

```python
# Create and access a nested dictionary

users = {
    "john": {"age": 30, "city": "New York"},
    "ani": {"age": 34, "city": "Philly"},
}
print(users["ani"])
print(users["ani"]["city"])
```

```python
# Change a value and add a new key in a nested dictionary

users["john"]["age"] = 31
users["ani"]["email"] = "ani@example.com"
print(users)
```

```python
# Iterate over a nested dictionary

for name, info in users.items():
    for key, value in info.items():
        print(name, key, value)
```

```python
# Accessing a missing outer key with [] raises an error

try:
    print(users["max"]["city"])
except Exception as ex:
    traceback.print_exc()
```

```python
# Chaining get() with an empty dictionary as the default avoids the error

print(users.get("max", {}).get("city"))
print(users.get("max", {}).get("city", "unknown"))
```

### defaultdict and Counter

```python
# defaultdict creates a default value for a missing key, which is useful for grouping

from collections import defaultdict

words = ["apple", "avocado", "banana", "cherry", "blueberry"]
groups = defaultdict(list)
for word in words:
    groups[word[0]].append(word)
print(dict(groups))
```

```python
# Counter is a dictionary that counts how many times each item appears

from collections import Counter

counts = Counter("mississippi")
print(counts)
print(counts["s"])
print(counts["z"])
print(counts.most_common(2))
```
