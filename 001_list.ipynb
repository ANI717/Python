```python
print(mylist[0])
print(mylist[1])
print(mylist[2])
```

```python
print(mylist[3])
```

```python
print(mylist[-1])
print(mylist[-2])
print(mylist[-3])
```

```python
print(mylist[-4])
```

#### Iterate

```python
for item in mylist:
    print(item)
```

#### Check item

```python
print("banana" in mylist)
print("grape" in mylist)
```

```python
if "banana" in mylist:
    print("Banana is in the list")
else:
    print("Banana is not in the list")

if "grape" in mylist:
    print("Grape is in the list")
else:
    print("Grape is not in the list")
```
