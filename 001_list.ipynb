```python
# Demonstrating the creation and printing of an empty list

empty_list = []
print(empty_list)
```

```python
# Demonstrating the creation and printing of a basic list

basic_list = ["banana", "apple", "pinaple"]
print(basic_list)
```

```python
# Demonstrating the creation and printing of a mixed-type list

mixed_list = ["banana", 1, True, 3.14]
print(mixed_list)
```

```python
# Demonstrating the creation and printing of a list with duplicate elements

duplicate_element_list = ["apple", "banana", "apple", "orange"]
print(duplicate_element_list)
```

```python
# Demonstrating the creation and printing of a list

mylist = ["banana", "apple", "orange"]
print(mylist)
```

### Indexing

```python
# Demonstrating accessing elements of a list by index

print(mylist[0])
print(mylist[1])
print(mylist[2])
```

```python
# Demonstrating accessing an element of a list by an index that is out of range

print(mylist[3])
```

```python
# Demonstrating accessing elements of a list using negative indices

print(mylist[-1])
print(mylist[-2])
print(mylist[-3])
```

```python
# Demonstrating accessing an element of a list using a negative index that is out of range

print(mylist[-4])
```

### Iterate

```python
# Demonstrating iterating over a list using a for loop

for item in mylist:
    print(item)
```

### Check item

```python
# Demonstrating checking for membership in a list

print("banana" in mylist)
print("grape" in mylist)
```

```python
# Demonstrating conditional statements based on membership in a list

if "banana" in mylist:
    print("Banana is in the list")
else:
    print("Banana is not in the list")

if "grape" in mylist:
    print("Grape is in the list")
else:
    print("Grape is not in the list")
```

### Length

```python
# Demonstrating getting the length of a list

print(len(mylist))
```

### Add item

```python
# Demonstrating appending an element to a list

mylist.append("pineapple")
print(mylist)
```

```python
# Demonstrating inserting an element at a specific position in a list

mylist.insert(2, "mango")
print(mylist)
```

### Remove item

```python
# Demonstrating removing the last element from a list using pop

item = mylist.pop()
print(item)
print(mylist)
```

```python
# Demonstrating removing a specific element from a list using remove

mylist.remove("apple")
print(mylist)
```

```python
# Demonstrating removing another specific element from a list using remove

mylist.remove("cherry")
```

```python
# Demonstrating clearing all elements from a list

mylist.clear()
print(mylist)
```

### Ordering

```python
# Demonstrating creating and printing a list

mylist = ["banana", "apple", "orange"]
print(mylist)
```

```python
# Demonstrating reversing the order of elements in a list

mylist.reverse()
print(mylist)
```

```python
# Demonstrating sorting the elements of a list in ascending order

mylist.sort()
print(mylist)
```

```python
# Demonstrating sorting a mixed list (if it contains comparable elements)

mixed_list.sort()
```

```python
# Demonstrating the use of the sorted() function to sort a list without modifying the original list

original_list = [1, 2, 3, -1, 0, 5,]
sorted_list = sorted(original_list)
print(original_list)
print(sorted_list)
```
