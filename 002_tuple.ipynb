```python
import sys
import timeit
import traceback
```

### Tuple

```python
# Creating and printing a tuple with multiple elements

tuple1 = (1, 2, 3, 4, 5)
print(tuple1)
print(type(tuple1))
```

```python

# Creating and printing a tuple without parentheses

tuple2 = 1, 2, 3, 4, 5
print(tuple2)
print(type(tuple2))
```

```python
# Creating and printing a tuple with a single element

tuple3 = (1,)
print(tuple3)
print(type(tuple3))
```

```python

# Creating and printing a tuple with a single element without parentheses

tuple4 = 1,
print(tuple4)
print(type(tuple4))
```

```python

# Creating and printing a tuple from a list

tuple5 = tuple([1, 2, 3])
print(tuple5)
print(type(tuple5))
```

```python
# Creating and printing a value that is not a tuple

not_a_tuple = (1)
print(not_a_tuple)
print(type(not_a_tuple))
```

```python
# Creating and printing an empty tuple

empty1 = ()
empty2 = tuple()
print(empty1, empty2)
print(type(empty1))
```

```python
# There is no tuple comprehension: parentheses around a comprehension create a generator

gen = (x * x for x in range(5))
print(type(gen))

squares = tuple(x * x for x in range(5))
print(squares)
print(type(squares))
```

### Indexing

```python
mytuple = (10, "goat", True, True, "extra")
print(mytuple)
print(type(mytuple))
```

```python
# Accessing and printing individual elements of the tuple

print(mytuple[0])
print(mytuple[1])
print(mytuple[2])
```

```python
# Accessing and printing elements using negative indexing

print(mytuple[-1])
print(mytuple[-2])
print(mytuple[-3])
```

### Change Item (not possible)

```python
# Attempting to modify a tuple element and handling the exception

try:
    mytuple[0] = "new_value"
except Exception as e:
    traceback.print_exc()
```

```python
# Attempting to delete a tuple element and handling the exception

try:
    del mytuple[0]
except Exception as e:
    traceback.print_exc()
```

```python
# Tuples have no methods that change them, such as append, remove or sort

try:
    mytuple.append("new_value")
except Exception as e:
    traceback.print_exc()
```

```python
# To "change" a tuple, convert it to a list, change the list and convert it back

colors = ("red", "green", "blue")
temp = list(colors)
temp[1] = "yellow"
temp.append("purple")
colors = tuple(temp)
print(colors)
```

### Immutable but not frozen

```python
# A tuple cannot change, but a mutable object inside it can

t = ([1, 2], 3)
t[0].append(3)
print(t)
```

```python
# Using += on a list inside a tuple raises an error, but the list is still changed

t = ([1, 2], 3)
try:
    t[0] += [4]
except Exception as e:
    traceback.print_exc()
print(t)
```

### Adding items creates a new tuple

```python
# Concatenating tuples creates a new tuple

t = (1, 2, 3)
print(t + (4, 5))
print(t * 2)
print(t)
```

```python
# += on a tuple creates a new object, but += on a list changes the same object

t = (1, 2, 3)
print(id(t))
t += (4,)
print(id(t), t)

l = [1, 2, 3]
print(id(l))
l += [4]
print(id(l), l)
```

### Unpack

```python
# Unpacking a tuple into separate variables

mytuple = "Ani", "34", "Philly"
name, age, city = mytuple
print(name)
print(age)
print(city)
```

```python
# Unpacking a tuple with the first and last elements separated from the rest

mytuple = (1, 2, 3, 4, 5, 6, 7)
num_first, *num_rest, num_last = mytuple
print(num_first)
print(num_rest)
print(num_last)
```

```python
# Swapping two variables using tuple packing and unpacking

a, b = 1, 2
a, b = b, a
print(a, b)
```

```python
# Returning multiple values from a function packs them into a tuple

def min_max(numbers):
    return min(numbers), max(numbers)

result = min_max([4, 1, 7, 3])
print(result)
print(type(result))

low, high = min_max([4, 1, 7, 3])
print(low, high)
```

### Hashable

```python
# A tuple can be used as a dictionary key or a set member

locations = {(39.95, -75.16): "Philly", (40.71, -74.01): "New York"}
print(locations[(39.95, -75.16)])

points = {(1, 2), (3, 4), (1, 2)}
print(points)
```

```python
# A list cannot be used as a dictionary key because it is not hashable

try:
    locations = {[39.95, -75.16]: "Philly"}
except Exception as e:
    traceback.print_exc()
```

```python
# A tuple that contains a list is not hashable either

try:
    print(hash((1, [2, 3])))
except Exception as e:
    traceback.print_exc()
```

### Named tuple

```python
# Demonstrating a named tuple, whose elements can be accessed by name or by index

from collections import namedtuple

Person = namedtuple("Person", ["name", "age", "city"])
person = Person("Ani", 34, "Philly")
print(person)
print(person.name)
print(person[1])
```

```python
# A named tuple is still immutable

try:
    person.age = 35
except Exception as e:
    traceback.print_exc()
```

### Tuple vs List

```python
# Comparing the memory size of a tuple and a list containing the same elements

mytuple = (10, "Ani", 30, True, "Philly")
mylist = [10, "Ani", 30, True, "Philly"]

print(sys.getsizeof(mytuple))
print(sys.getsizeof(mylist))
```

```python
# Comparing the time it takes to create a tuple and a list containing the same elements

print(timeit.timeit('mytuple = (10, "Ani", 30, True, "Philly")', number=1000000))
print(timeit.timeit('mylist = [10, "Ani", 30, True, "Philly"]', number=1000000))
```

```python
# A tuple literal is stored as a ready-made constant, while a list is built every time
# This is a big part of why the tuple was faster to create above

import dis

dis.dis('(10, "Ani", 30, True, "Philly")')
print()
dis.dis('[10, "Ani", 30, True, "Philly"]')
```

```python
# sorted() on a tuple returns a list

numbers = (3, 1, 2)
print(sorted(numbers))
print(tuple(sorted(numbers)))
```
