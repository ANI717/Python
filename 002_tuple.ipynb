```python
import sys
import timeit
import traceback
```

### Tuple

```python
# Creating and printing a tuple with multiple elements

tuple1 = (1, 2, 3, 4, 5)
print(tuple1)
print(type(tuple1))
```

```python

# Creating and printing a tuple without parentheses

tuple2 = 1, 2, 3, 4, 5
print(tuple2)
print(type(tuple2))
```

```python
# Creating and printing a tuple with a single element

tuple3 = (1,)
print(tuple3)
print(type(tuple3))
```

```python

# Creating and printing a tuple with a single element without parentheses

tuple4 = 1,
print(tuple4)
print(type(tuple4))
```

```python

# Creating and printing a tuple from a list

tuple5 = tuple([1, 2, 3])
print(tuple5)
print(type(tuple5))
```

```python
# Creating and printing a value that is not a tuple

not_a_tuple = (1)
print(not_a_tuple)
print(type(not_a_tuple))
```

### Indexing

```python
mytuple = (10, "goat", True, True, "extra")
print(mytuple)
print(type(mytuple))
```

```python
# Accessing and printing individual elements of the tuple

print(mytuple[0])
print(mytuple[1])
print(mytuple[2])
```

```python
# Accessing and printing elements using negative indexing

print(mytuple[-1])
print(mytuple[-2])
print(mytuple[-3])
```

### Change Item (not possible)

```python
# Attempting to modify a tuple element and handling the exception

try:
    mytuple[0] = "new_value"
except Exception as e:
    traceback.print_exc()
```

### Length

```python
# Printing the length of the tuple

print(len(mytuple))
```

### Count

```python
# Counting the number of True values in the tuple

print(mytuple.count(True))
```

### Index

```python
# Finding the index of the first True value in the tuple

print(mytuple.index(True))
```

### Slicing

```python
# Slicing the tuple to get specific ranges of elements

print(mytuple[1:3])
print(mytuple[:3])
print(mytuple[2:])
```

### Unpack

```python
# Unpacking a tuple into separate variables

mytuple = "Ani", "34", "Philly"
name, age, city = mytuple
print(name)
print(age)
print(city)
```

```python
# Unpacking a tuple with the first and last elements separated from the rest

mytuple = (1, 2, 3, 4, 5, 6, 7)
num_first, *num_rest, num_last = mytuple
print(num_first)
print(num_rest)
print(num_last)
```

### Tuple vs List

```python
# Comparing the memory size of a tuple and a list containing the same elements

mytuple = (10, "Ani", 30, True, "Philly")
mylist = [10, "Ani", 30, True, "Philly"]

print(sys.getsizeof(mytuple))
print(sys.getsizeof(mylist))
```

```python
# Comparing the time it takes to create a tuple and a list containing the same elements

print(timeit.timeit('mytuple = (10, "Ani", 30, True, "Philly")', number=1000000))
print(timeit.timeit('mylist = [10, "Ani", 30, True, "Philly"]', number=1000000))
```
